```python
# Sleep Quality Features
train['toplam_uyku_yuzdesi'] = train['rem_yuzdesi'] + train['derin_uyku_yuzdesi']
train['uyku_kalitesi_orani'] = train['toplam_uyku_yuzdesi'] / train['uykuya_dalma_suresi_dk']
train['gece_bolunme_skoru'] = train['gecelik_uyanma_sayisi'] * train['uykuya_dalma_suresi_dk']
train['stres_nabiz_etkilesimi'] = train['stres_skoru'] * train['dinlenik_nabiz_bpm']
train['saglik_skoru'] = train['gunluk_adim_sayisi'] / (train['stres_skoru'] + 1)
train['kafein_ekran_etkilesimi'] = train['uyku_oncesi_kafein_mg'] * train['uyku_oncesi_ekran_suresi_dk']
train['yasam_dengesi'] = train['gunluk_calisma_saati'] / (train['sekerleme_suresi_dk'] + 1)

# Apply same to test
test['toplam_uyku_yuzdesi'] = test['rem_yuzdesi'] + test['derin_uyku_yuzdesi']
test['uyku_kalitesi_orani'] = test['toplam_uyku_yuzdesi'] / test['uykuya_dalma_suresi_dk']
test['gece_bolunme_skoru'] = test['gecelik_uyanma_sayisi'] * test['uykuya_dalma_suresi_dk']
test['stres_nabiz_etkilesimi'] = test['stres_skoru'] * test['dinlenik_nabiz_bpm']
test['saglik_skoru'] = test['gunluk_adim_sayisi'] / (test['stres_skoru'] + 1)
test['kafein_ekran_etkilesimi'] = test['uyku_oncesi_kafein_mg'] * test['uyku_oncesi_ekran_suresi_dk']
test['yasam_dengesi'] = test['gunluk_calisma_saati'] / (test['sekerleme_suresi_dk'] + 1)

print(train.shape, test.shape)
# Cap AFTER both train and test features are created
for col in ['kafein_ekran_etkilesimi', 'gece_bolunme_skoru', 'uyku_kalitesi_orani']:
    upper = train[col].quantile(0.99)
    train[col] = train[col].clip(upper=upper)
    test[col] = test[col].clip(upper=upper)

print(train.shape, test.shape)
```

# Encoding

```python
from sklearn.model_selection import KFold
import numpy as np

# Binary encoding
binary_map = {
    'cinsiyet': {'Kadin': 0, 'Erkek': 1},
    'mevsim': {'Ilkbahar-Yaz': 0, 'Sonbahar-Kis': 1},
    'gun_tipi': {'Hafta ici': 0, 'Hafta sonu': 1}
}

# Ordinal encoding
ordinal_map = {
    'kronotip': {'Bilinmiyor': -1, 'Sabah insani': 0, 'Notr': 1, 'Gece insani': 2},
    'ruh_sagligi_durumu': {'Bilinmiyor': -1, 'Saglikli': 0, 'Anksiyete': 1, 'Depresyon': 2, 'Anksiyete ve depresyon': 3}
}

for col, mapping in {**binary_map, **ordinal_map}.items():
    train[col] = train[col].map(mapping)
    test[col] = test[col].map(mapping)

# Target encoding for meslek and ulke using KFold to avoid leakage
kf = KFold(n_splits=5, shuffle=True, random_state=42)

for col in ['meslek', 'ulke']:
    train[f'{col}_target'] = 0.0
    for train_idx, val_idx in kf.split(train):
        mean_map = train.iloc[train_idx].groupby(col)['bilissel_performans_skoru'].mean()
        train.loc[train.index[val_idx], f'{col}_target'] = train.iloc[val_idx][col].map(mean_map)
    
    # For test use full train mean
    test[f'{col}_target'] = test[col].map(train.groupby(col)['bilissel_performans_skoru'].mean())
    
    # Drop original column
    train = train.drop(columns=[col])
    test = test.drop(columns=[col])

print(train.shape, test.shape)
print(train.dtypes.value_counts())
```
